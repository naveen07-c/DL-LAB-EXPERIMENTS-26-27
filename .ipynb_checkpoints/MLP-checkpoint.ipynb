```python
class MLP:
    def __init__(self,input_size,hidden_size, output_size,lr=0.1):
        self.lr =lr

        #layer 1  input -> hidden layer
        self.w1= np.random.randn(input_size,hidden_size)
        self.b1 = np.zeros((1,hidden_size))

        #layer 2 hidden layer ->output layer
        self.w2=np.random.randn(hidden_size,output_size)
        self.b2 =np.zeros((1,output_size))

    def sigmoid(self,x):
        return 1/(1+np.exp(-x))

    def sigmoid_derivative(self,x):
        return x * (1 - x)

    def forward(self,x):
        # forward of the first layer (input --> hidden layer)
        self.z1=np.dot(x,self.w1)+ self.b1
        self.a1=self.sigmoid(self.z1)


        # forward pass of hidden ---> output layer
        
        self.z2=np.dot(self.a1,self.w2)+self.b2
        self.a2=self.sigmoid(self.z2)

        # a2 is the output of the network which is the result of the forward pass
        return self.a2

    def backward(self,x,y):

        # calculate the loss which is  y - y_pred  where y_pred is the output of the network
        error= y - self.a2

        d_output = error * self.sigmoid_derivative(self.a2)

        hidden_error = np.dot(d_output, self.w2.T)
        d_hidden = hidden_error * self.sigmoid_derivative(self.a1)

        self.w2 += self.lr * np.dot(self.a1.T, d_output)
        self.b2 += self.lr * np.sum(d_output, axis=0, keepdims=True)

        self.w1 += self.lr * np.dot(X.T, d_hidden)
        self.b1 += self.lr * np.sum(d_hidden, axis=0, keepdims=True)

        return np.mean(error ** 2)

    def train(self, X, y, epochs=20000):
        for epoch in range(epochs):
            self.forward(X)
            loss = self.backward(X, y)

            if epoch % 1000 == 0:
                print(f"Epoch {epoch}, Loss: {loss:.6f}")

    def predict(self, X):
        return np.round(self.forward(X)) 
```
